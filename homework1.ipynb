```python
import pandas as pd
import numpy as np
print("pandas version:",pd.__version__)
url ='https://raw.githubusercontent.com/alexeygrigorev/datasets/master/car_fuel_efficiency.csv'
df = pd.read_csv(url)
df.head()
```

```python
median_before = df['horsepower'].median()
mode_hp = df['horsepower'].mode()[0]
df_filled = df['horsepower'].copy()
df_filled = df_filled.fillna(mode_hp)
median_after = df_filled.median()
print("before:", median_before)
print("after:", median_after)
```

```python
features = ['engine_displacement', 'horsepower', 'acceleration']
X = df[features].fillna(0).values
y = df['fuel_efficiency_mpg'].values
ones = np.ones(X.shape[0])
X = np.column_stack([ones, X])
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
w = XTX_inv.dot(X.T).dot(y)
print("Sum of weights:", w.sum())
```
